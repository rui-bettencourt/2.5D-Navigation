```python
# SuperFastPython.com
# example of a sequential for loop
from time import sleep
from random import random
 
# a task to execute in another thread
def task(arg):
    # generate a value between 0 and 1
    value = random()
    # block for a fraction of a second to simulate work
    sleep(value)
    # return the generated value
    return value
 
# entry point for the program
if __name__ == '__main__':
    # call the same function with different data sequentially
    for result in map(task, range(10)):
        # report the value to show progress
        print(result)
```

```python
# SuperFastPython.com
# example of a concurrent for loop
from time import sleep
from random import random
import numpy as np
from concurrent.futures import ThreadPoolExecutor

class We(object):
    # task to execute in another thread
    def sum(self, arg1, arg2):
        return self.ber + arg1 + arg2
    def task(self, arg1,arg2):
        result = np.asarray([self.ber, self.sum(arg1, arg2)])
        # return the generated value
        return result

    def run(self):
        # create the thread pool
        self.ber = 5
        with ThreadPoolExecutor() as tpe:
            # call the same function with different data concurrently
            result = list(tpe.map(self.task, [0,1],[2,3]))
            result = np.vstack(result)
            # report the value to show progress
            print(result)
            print(result.sum(axis=0))
# entry point for the program
if __name__ == '__main__':
    a = We()
    a.run()
```
